## 4. Basic EDA

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Machine failure")

plt.title("Machine Failure Count")
plt.xlabel("Machine Failure")
plt.ylabel("Count")
plt.xticks([0, 1], ["No Failure", "Failure"])

plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.countplot(data=df, x="Type", hue="Machine failure")

plt.title("Machine Failure by Product Type")
plt.xlabel("Product Type")
plt.ylabel("Count")
plt.legend(title="Machine Failure", labels=["No Failure", "Failure"])

plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(data=df, x="Machine failure", y="Tool wear [min]")

plt.title("Tool Wear Distribution by Machine Failure")
plt.xlabel("Machine Failure")
plt.ylabel("Tool Wear [min]")
plt.xticks([0, 1], ["No Failure", "Failure"])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Rotational speed [rpm]",
    y="Torque [Nm]",
    hue="Machine failure",
    alpha=0.6
)

plt.title("Torque vs Rotational Speed by Machine Failure")
plt.xlabel("Rotational Speed [rpm]")
plt.ylabel("Torque [Nm]")
plt.legend(title="Machine Failure", labels=["No Failure", "Failure"])

plt.show()
```

```python
numeric_cols = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
    "Machine failure",
    "TWF",
    "HDF",
    "PWF",
    "OSF",
    "RNF"
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Sensor Variables and Failure Labels")

plt.show()
```

## 5. Initial Findings

## Initial Findings

1. The dataset contains 10,000 production records and 14 variables, including product information, operating conditions, sensor measurements, a machine failure label, and several specific failure type indicators.

2. There are no missing values in the dataset, which makes it suitable for exploratory data analysis and machine learning modeling without major data cleaning at this stage.

3. The target variable `Machine failure` is highly imbalanced. Most records represent normal machine operation, while only a small portion of records are failure cases. This means that accuracy alone will not be a reliable metric in the later modeling stage. Precision, recall, F1-score, and confusion matrix analysis will be more important for evaluating whether the model can actually detect failure cases.

4. Product type distribution shows that type `L` has the largest number of records and also the largest number of failure cases in absolute count. However, because the three product types have different sample sizes, failure count alone is not enough to conclude which product type is riskier. A failure rate by product type should be calculated in the next step.

5. The boxplot of `Tool wear [min]` shows that failure cases tend to have a higher median tool wear than non-failure cases. This suggests that accumulated tool usage is likely an important factor in machine failure risk, especially for wear-related failures.

6. The scatter plot of `Torque [Nm]` versus `Rotational speed [rpm]` shows a strong inverse relationship between torque and rotational speed. Failure cases appear more concentrated in extreme operating regions, especially high-torque / low-speed conditions and very high-speed / low-torque conditions. This suggests that machine failure may depend on combinations of operating variables rather than one variable alone.

7. The correlation heatmap shows that `Air temperature [K]` and `Process temperature [K]` are strongly correlated, with a correlation of approximately 0.88. This is expected because process temperature is partly influenced by the surrounding air temperature. In later modeling, feature engineering such as `Temperature difference = Process temperature - Air temperature` may be more informative than using both temperatures independently.

8. `Rotational speed [rpm]` and `Torque [Nm]` have a strong negative correlation of approximately -0.88. This indicates that the machine often operates under a trade-off between speed and load. Because mechanical power is related to torque and rotational speed, a new feature such as `Power_proxy = Torque × Rotational speed` should be created in the feature engineering stage.

9. Among the operating variables, `Torque [Nm]` and `Tool wear [min]` show the clearest positive correlations with `Machine failure`, although the correlations are still moderate. This suggests that machine failure is likely driven by nonlinear interactions among multiple variables, rather than by a single feature.

10. The specific failure type indicators show strong relationships with the overall `Machine failure` label, especially `HDF`, `PWF`, and `OSF`. This confirms that the dataset contains multiple failure mechanisms, including heat dissipation failure, power failure, and overstrain failure. In the first modeling stage, the project will focus on binary failure prediction. Later, the analysis can be extended to failure type classification.

11. Based on the initial EDA, the next steps are to create engineered features such as temperature difference and power proxy, calculate failure rates by product type, and build baseline classification models to predict machine failure.
